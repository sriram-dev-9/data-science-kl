```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(
    {
        "Student Name" : ["Nikhil","Sreyas","Pardheev"],
        "Age" : [20,np.nan,19],
        "Gender": ['M',np.nan,'M'],
        "Department": ['CSE','CSE',np.nan]
    }
    )
```

```python
df
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
df
```

```python
df['Department'] = df['Department'].ffill()
df
```

```python
df['Gender'] = df['Gender'].bfill()
df
```
